# Importation des bibliothèques


```python
import numpy as np
import matplotlib.pyplot as plt 
import scipy as sp 
import pandas as pd 
from tkinter import filedialog 
import glob
import os
```

```python
spectres = filedialog.askopenfilenames(title="Selectionnez le fichier CSV", filetypes=[("csv files", "*.csv , *.CSV , *.txt")])
df1 = pd.read_csv(spectres[0], sep=',', decimal='.', header=0, skiprows=0, encoding='utf-8')
df1.head()
```

```python
# l’axe X doit correspondre au nombre de colonnes du DataFrame

n_points = 2090
x = np.linspace(100, 4278, n_points)

plt.figure(figsize=(8, 5), dpi=200)

# tracer chaque ligne (chaque spectre)
for i in range(df1.shape[0]):              
    y = df1.iloc[i, :].astype(float).values  
    plt.plot(x,y, alpha=0.3,color="green") 

    
plt.xlabel("longueur d'onde (cm-1)")    
plt.ylabel("Intensité")
plt.title("Visuel des spectres Raman des cellules cancereuses")
plt.plot([], [], color="green", label="(COOH)2") #Ajout de 2 lignes fictives pour afficher les légendes
plt.legend()
plt.grid(True)
plt.show()
```

> Dataset : Raman spectra of the culture medium of normal and cancer cells
